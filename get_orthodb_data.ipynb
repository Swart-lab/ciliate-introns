Which BUSCO markers are associated with spliceosome?

Mesodinium rubrum genome has low BUSCO completeness score. We think a possible reason for this is that BUSCO conserved markers include spliceosomal complex components, but Mesodinium appears to lack a spliceosome.

We therefore want to check which BUSCO markers (Alveolata, ODB_10 marker set) are spliceosome-associated.

Get data from OrthoDB API. Unfortunately the SPARQL query interface is only available for the latest OrthoDB version (v11), but the latest BUSCO marker set is based on OrthoDB v10. There does not appear to be a mapping between orthogroups between different versions.

```python
import requests
```

```python
url = "https://v10-1.orthodb.org/group"
```

```python
import pandas as pd
```

```python
df = pd.read_csv("/mnt/data/db/busco/lineages/alveolata_odb10/links_to_ODB10.txt", sep="\t", names=['odbgroup','label','url'])
```

```python
df.head()
```

```python
ids = list(df['odbgroup'])
odb_data = {}
```

```python
for i in ids:
    print(i)
    r = requests.get(url, params={'id':i})
    if r.ok:
        odb_data[i] = r.json()['data']
```

```python
len(odb_data)
```

```python
for i in odb_data:
    if 'KEGGpathway' in odb_data[i] and any([j['description'] == 'Spliceosome' for j in odb_data[i]['KEGGpathway']]):
        print(i + "\t" + odb_data[i]['name'])
```

```python
for i in odb_data:
    if 'cellular_component' in odb_data[i] and any([j['id'] == 'GO:0005681' for j in odb_data[i]['cellular_component']]):
        print(i + "\t" + odb_data[i]['name'])
```
